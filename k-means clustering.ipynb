```python
import os
os.environ["OMP_NUM_THREADS"] = "1"
```

```python
# Step 1: Import libraries
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

```

```python
# Step 2: Load the dataset
iris = load_iris()
X = iris.data[:, :2]  # Sepal length and sepal width

print("Dataset shape:", X.shape)
```

```python
# Step 3: Standardize the data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

```

```python

# Step 4: Apply the Elbow Method
inertia = []

for k in range(1, 11):
    model = KMeans(n_clusters=k, random_state=42, n_init=10)
    model.fit(X_scaled)
    inertia.append(model.inertia_)
```

```python
# Step 5: Plot the Elbow graph
plt.plot(range(1, 11), inertia, marker='o')
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Method")
plt.show()

```

```python

# Step 6: Apply K-Means with K = 3
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
clusters = kmeans.fit_predict(X_scaled)

```

```python
# Step 7: Visualize the clusters
plt.scatter(
    X_scaled[:, 0], X_scaled[:, 1],
    c=clusters, cmap='viridis', s=50
)

plt.scatter(
    kmeans.cluster_centers_[:, 0],
    kmeans.cluster_centers_[:, 1],
    marker='X', s=200, color='red', label='Centroids'
)

plt.xlabel("Standardized Sepal Length")
plt.ylabel("Standardized Sepal Width")
plt.title("K-Means Clustering (K = 3)")
plt.legend()
plt.show()
```

```python
# Step 8: Display cluster assignments
print("\nFirst 10 cluster assignments:")
print(clusters[:10])

```
