```python
import numpy as np
import pandas as pd

# Create data using NumPy
data = np.array([
    [101, "Alice", 85],
    [102, "Bob", 90],
    [103, "Charlie", 78],
    [104, "David", 92],
    [105, "Emma", 88]
])

# Create a Pandas DataFrame
df = pd.DataFrame(data, columns=["ID", "Name", "Marks"])

print("Original DataFrame:")
print(df)

# ---------------- INDEXING ----------------

# Select a single column
print("\nName column:")
print(df["Name"])

# Select multiple columns
print("\nName and Marks columns:")
print(df[["Name", "Marks"]])

# Select a row using index
print("\nFirst row:")
print(df.iloc[0])

# Select a specific value
print("\nMarks of Alice:")
print(df.iloc[0, 2])

# ---------------- SLICING ----------------

# Slice rows from index 1 to 3
print("\nRows 1 to 3:")
print(df.iloc[1:4])

# Slice first three rows
print("\nFirst three rows:")
print(df.iloc[:3])

# Slice last two rows
print("\nLast two rows:")
print(df.iloc[-2:])

# Slice rows and columns
print("\nRows 1 to 3 and columns Name to Marks:")
print(df.iloc[1:4, 1:3])
```
